```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Input directory
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Input_Data/Imputed_Data_NSW_ALL'
output_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Output_Plots'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Define PM2.5 categories based on NSW AQI standards
def categorize_pm25(pm_value):
    if pm_value <= 8:
        return 'Good'
    elif 9 <= pm_value <= 16:
        return 'Fair'
    elif 17 <= pm_value <= 25:
        return 'Poor'
    elif 26 <= pm_value <= 50:
        return 'Very Poor'
    else:
        return 'Hazardous'

# Function to determine the season based on the month
def get_season(month):
    if month in [12, 1, 2]:
        return 'Summer'
    elif month in [3, 4, 5]:
        return 'Autumn'
    elif month in [6, 7, 8]:
        return 'Winter'
    else:
        return 'Spring'

# Initialize an empty DataFrame to store all categorized data
all_data = []

# Iterate over all CSV files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith('.csv'):
        # Extract StudySite (first word of the filename)
        study_site = filename.split('_')[0]
        
        # Read the CSV file
        file_path = os.path.join(input_directory, filename)
        df = pd.read_csv(file_path)
        
        # Check if PM2.5 column exists and the Date column is available
        if 'PM2.5' in df.columns and 'Date' in df.columns:
            # Convert Date to datetime format, with error handling
            df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
            
            # Drop rows where Date is NaT
            df = df.dropna(subset=['Date'])
            
            # Categorize PM2.5 values
            df['Category'] = df['PM2.5'].apply(categorize_pm25)
            df['StudySite'] = study_site
            
            # Extract the season from the month
            df['Season'] = df['Date'].dt.month.apply(get_season)
            # Extract hour from the Date
            df['Hour'] = df['Date'].dt.hour
            
            # Append the relevant columns to the all_data list
            all_data.append(df[['StudySite', 'Category', 'Season', 'Hour']])

# Concatenate all data into a single DataFrame
combined_data = pd.concat(all_data)

# Filter data for the Hazardous category
hazardous_data = combined_data[combined_data['Category'] == 'Hazardous']

# Get unique StudySites
study_sites = hazardous_data['StudySite'].unique()

# Generate a plot for each StudySite focusing on Hazardous PM2.5 values by hour
for site in study_sites:
    plt.figure(figsize=(12, 6))
    
    # Filter data for the current StudySite
    site_data = hazardous_data[hazardous_data['StudySite'] == site]
    
    # Calculate counts of Hazardous occurrences by hour and season
    count_data = site_data.groupby(['Hour', 'Season']).size().unstack(fill_value=0)
    
    # Plotting
    count_data.plot(kind='bar', stacked=False, color=['red', 'orange', 'blue', 'green'], ax=plt.gca())
    plt.title(f'Hourly Distribution of Hazardous PM2.5 Values by Season for {site}')
    plt.ylabel('Count')
    plt.xlabel('Hour of the Day')
    plt.xticks(rotation=0)
    plt.legend(title='Season')
    plt.tight_layout()
    
    # Save the plot
    plot_filename = os.path.join(output_directory, f'{site}_Hazardous_PM2.5_Hourly_By_Season.png')
    plt.savefig(plot_filename)
    
    plt.close()

print(f"All plots for hourly distribution of Hazardous PM2.5 values by season have been generated and saved in {output_directory}")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Input directory
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Input_Data/Imputed_Data_NSW_ALL'
output_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Output_Plots45'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Define PM2.5 categories based on NSW AQI standards
def categorize_pm25(pm_value):
    if pm_value <= 8:
        return 'Good'
    elif 9 <= pm_value <= 16:
        return 'Fair'
    elif 17 <= pm_value <= 25:
        return 'Poor'
    elif 26 <= pm_value <= 50:
        return 'Very Poor'
    else:
        return 'Hazardous'

# Function to determine the season based on the month
def get_season(month):
    if month in [12, 1, 2]:
        return 'Summer'
    elif month in [3, 4, 5]:
        return 'Autumn'
    elif month in [6, 7, 8]:
        return 'Winter'
    else:
        return 'Spring'

# Initialize an empty DataFrame to store all categorized data
all_data = []

# Iterate over all CSV files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith('.csv'):
        # Extract StudySite (first word of the filename)
        study_site = filename.split('_')[0]
        
        # Read the CSV file
        file_path = os.path.join(input_directory, filename)
        df = pd.read_csv(file_path)
        
        # Check if PM2.5 column exists and the Date column is available
        if 'PM2.5' in df.columns and 'Date' in df.columns and 'Time' in df.columns:
            # Convert Date to datetime format, with error handling
            df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
            
            # Drop rows where Date is NaT
            df = df.dropna(subset=['Date'])
            
            # Categorize PM2.5 values
            df['Category'] = df['PM2.5'].apply(categorize_pm25)
            df['StudySite'] = study_site
            
            # Extract the season from the month
            df['Season'] = df['Date'].dt.month.apply(get_season)
            # Use the Time column directly and convert to hour format (0-23)
            df['Hour'] = df['Time'].apply(lambda x: x - 1 if x == 24 else x)
            
            # Append the relevant columns to the all_data list
            all_data.append(df[['StudySite', 'Category', 'Season', 'Hour']])

# Concatenate all data into a single DataFrame
combined_data = pd.concat(all_data)

# Filter data for the Hazardous category
hazardous_data = combined_data[combined_data['Category'] == 'Hazardous']

# Get unique StudySites
study_sites = hazardous_data['StudySite'].unique()

# Generate a subplot for each StudySite focusing on Hazardous PM2.5 values by hour
for site in study_sites:
    fig, axes = plt.subplots(2, 2, figsize=(12, 10), sharey=True)
    fig.suptitle(f'Hazardous PM2.5 Values Count by Hour for {site}', fontsize=16)

    # Filter data for the current StudySite
    site_data = hazardous_data[hazardous_data['StudySite'] == site]

    # Define the seasons
    seasons = ['Summer', 'Autumn', 'Winter', 'Spring']
    
    # For each season, create a count plot in a separate subplot
    for ax, season in zip(axes.flatten(), seasons):
        # Filter data for the current season
        season_data = site_data[site_data['Season'] == season]
        
        # Calculate counts of Hazardous occurrences by hour
        count_data = season_data.groupby('Hour').size().reset_index(name='Count')
        
        # Create the bar plot for counts
        ax.bar(count_data['Hour'], count_data['Count'], color='red')
        ax.set_title(season)
        ax.set_ylabel('Count')
        ax.set_xlabel('Hour of the Day')
        ax.set_xticks(range(1, 25))  # Set x-ticks from 1 to 24
        ax.set_xticklabels(range(1, 25))  # Label x-ticks from 1 to 24

    plt.tight_layout(rect=[0, 0, 1, 0.95])  # Adjust layout to make room for the title
    
    # Save the plot
    plot_filename = os.path.join(output_directory, f'{site}_Hazardous_PM2.5_By_Season_Count.png')
    plt.savefig(plot_filename)
    
    plt.close()

print(f"All plots for hourly distribution of Hazardous PM2.5 values by season have been generated and saved in {output_directory}")
```

```python
import os
import pandas as pd

# Input directory
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Input_Data/Imputed_Data_NSW_ALL'
output_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Output_Data_hazardous_values'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Function to categorize PM2.5 values based on NSW AQI standards
def categorize_pm25(pm_value):
    if pm_value <= 8:
        return 'Good'
    elif 9 <= pm_value <= 16:
        return 'Fair'
    elif 17 <= pm_value <= 25:
        return 'Poor'
    elif 26 <= pm_value <= 50:
        return 'Very Poor'
    else:
        return 'Hazardous'

# Initialize a dictionary to store DataFrames for each StudySite
study_site_data = {}

# Define the cutoff date
cutoff_date = pd.Timestamp('2020-04-30')

# Iterate over all CSV files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith('.csv'):
        # Read the CSV file
        file_path = os.path.join(input_directory, filename)
        df = pd.read_csv(file_path)
        
        # Check if PM2.5 column exists and the Date and Time columns are available
        if 'PM2.5' in df.columns and 'Date' in df.columns and 'Time' in df.columns:
            # Convert the Date column to datetime format
            df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
            
            # Drop rows where Date is NaT
            df = df.dropna(subset=['Date'])
            
            # Filter out data before the cutoff date
            df = df[df['Date'] >= cutoff_date]

            # Categorize PM2.5 values
            df['Category'] = df['PM2.5'].apply(categorize_pm25)
            
            # Log unexpected hazardous values
            unexpected_hazardous = df[(df['Category'] == 'Hazardous') & (df['PM2.5'] <= 50)]
            if not unexpected_hazardous.empty:
                print(f"Unexpected hazardous values found in {filename}:\n{unexpected_hazardous[['Date', 'Time', 'PM2.5']]}\n")

            # Filter for Hazardous values
            hazardous_df = df[df['Category'] == 'Hazardous']
            
            # Select relevant columns
            hazardous_values = hazardous_df[['Date', 'Time', 'PM2.5']]
            
            # Get the StudySite from the filename (first word)
            study_site = filename.split('_')[0]
            
            # Append to the corresponding StudySite DataFrame
            if study_site not in study_site_data:
                study_site_data[study_site] = hazardous_values
            else:
                study_site_data[study_site] = pd.concat([study_site_data[study_site], hazardous_values])

# Save each StudySite DataFrame to an individual CSV file
for study_site, data in study_site_data.items():
    csv_file_path = os.path.join(output_directory, f'Hazardous_PM2.5_Values_{study_site}.csv')
    data.to_csv(csv_file_path, index=False)

print(f"All hazardous PM2.5 values along with Date and Time after April 30, 2020, have been saved for each StudySite in {output_directory}")
```
