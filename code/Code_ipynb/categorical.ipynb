```python
import os
import pandas as pd

# Input directory
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Input_Data/Imputed_Data_NSW_ALL'

# Function to categorize PM2.5 values based on NSW AQI standards
def categorize_pm25(pm_value):
    if pm_value <= 8:
        return 'Good'
    elif 9 <= pm_value <= 16:
        return 'Fair'
    elif 17 <= pm_value <= 25:
        return 'Poor'
    elif 26 <= pm_value <= 50:
        return 'Very Poor'
    else:
        return 'Hazardous'

# Iterate over all CSV files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith('.csv'):
        # Read the CSV file
        file_path = os.path.join(input_directory, filename)
        df = pd.read_csv(file_path)

        # Check if PM2.5 column exists
        if 'PM2.5' in df.columns:
            # Categorize PM2.5 values
            df['Category'] = df['PM2.5'].apply(categorize_pm25)

            # Save the updated DataFrame back to the same CSV file
            df.to_csv(file_path, index=False)

print(f"Added PM2.5 category column to all CSV files in {input_directory}.")
```

```python
import os
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Input and output directories
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nowcasting/Input_Data/Imputed_Data_NSW_ALL'
output_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nowcasting/Input_Data_CAT_OneHot'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Function to categorize PM2.5 values based on detailed AQI standards
def categorize_pm25(pm_value):
    if pm_value <= 12.0:
        return 'Good'
    elif 12.1 <= pm_value <= 35.4:
        return 'Moderate'
    elif 35.5 <= pm_value <= 55.4:
        return 'Unhealthy for Sensitive Groups'
    elif 55.5 <= pm_value <= 150.4:
        return 'Unhealthy'
    elif 150.5 <= pm_value <= 200.0:
        return 'Very Unhealthy'
    elif 200.1 <= pm_value <= 250.0:
        return 'Very Very Unhealthy'
    elif 250.5 <= pm_value <= 350.0:
        return 'Hazardous'
    else:
        return 'Very Hazardous'

# Initialize OneHotEncoder
encoder = OneHotEncoder(sparse=False)

# Iterate over all CSV files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith('.csv'):
        # Read the CSV file
        file_path = os.path.join(input_directory, filename)
        df = pd.read_csv(file_path)

        # Check if PM2.5 column exists
        if 'PM2.5' in df.columns:
            # Categorize PM2.5 values
            df['Category'] = df['PM2.5'].apply(categorize_pm25)

            # Apply OneHotEncoder to the 'Category' column
            category_encoded = encoder.fit_transform(df[['Category']])
            category_encoded_df = pd.DataFrame(
                category_encoded,
                columns=encoder.get_feature_names(['Category'])  # Replace with get_feature_names_out if upgraded
            )

            # Concatenate the encoded categories with the original DataFrame
            df = pd.concat([df, category_encoded_df], axis=1)

            # Save the updated DataFrame to the output directory
            output_file_path = os.path.join(output_directory, filename)
            df.to_csv(output_file_path, index=False)

print(f"All updated CSV files with PM2.5 categories and OneHotEncoded categories have been saved in {output_directory}.")
```
