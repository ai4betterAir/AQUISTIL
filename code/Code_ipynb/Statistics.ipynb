```python
import os
import pandas as pd

# Input and output directories
input_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Input_Data/Imputed_Data_NSW_ALL'
output_directory = '/mnt/scratch_lustre/barthelx/Masrur/Nawcasting/Output_Data_Descriptive_Statistics'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Dictionary to hold maximum PM2.5 values for each StudySite
max_pm25_values = {}

# Initialize an Excel writer
excel_file_path = os.path.join(output_directory, 'Descriptive_Statistics_by_StudySite.xlsx')
with pd.ExcelWriter(excel_file_path) as writer:
    # Iterate over all CSV files in the input directory
    for filename in os.listdir(input_directory):
        if filename.endswith('.csv'):
            # Read the CSV file
            file_path = os.path.join(input_directory, filename)
            df = pd.read_csv(file_path)

            # Check if PM2.5 column exists
            if 'PM2.5' in df.columns:
                # Get the StudySite from the filename (first word)
                study_site = filename.split('_')[0]
                
                # Calculate descriptive statistics
                descriptive_stats = df.describe()
                
                # Save descriptive statistics to a separate sheet in the Excel file
                descriptive_stats.to_excel(writer, sheet_name=study_site)

                # Calculate the maximum PM2.5 value for the StudySite
                max_pm25 = df['PM2.5'].max()
                max_pm25_values[study_site] = max_pm25

# Create a DataFrame from the max PM2.5 values dictionary
max_pm25_df = pd.DataFrame(list(max_pm25_values.items()), columns=['StudySite', 'Max_PM2.5'])

# Save the maximum PM2.5 values to a CSV file
max_pm25_csv_path = os.path.join(output_directory, 'Max_PM2.5_Values.csv')
max_pm25_df.to_csv(max_pm25_csv_path, index=False)

print(f"Descriptive statistics have been saved to {excel_file_path}.")
print(f"Maximum PM2.5 values have been saved to {max_pm25_csv_path}.")
```
